# 04 · Evaluación (OE4 · IND-06)

**Metas:** F1 ≥ 0.75 y AUC-ROC > 0.80 sobre el conjunto de prueba (sin SMOTE).

```python
import pandas as pd
import joblib
import matplotlib.pyplot as plt
from sklearn.metrics import (
    classification_report, ConfusionMatrixDisplay, RocCurveDisplay,
    PrecisionRecallDisplay, f1_score, roc_auc_score,
)

test = pd.read_csv("../data/test.csv")
X_te, y_te = test.drop(columns="en_riesgo"), test["en_riesgo"]
modelo = joblib.load("../../ml-service/model/model.joblib")
pred = modelo.predict(X_te)
proba = modelo.predict_proba(X_te)[:, 1]
```

```python
print(classification_report(y_te, pred, digits=3))
f1, auc = f1_score(y_te, pred), roc_auc_score(y_te, proba)
print(f"F1 = {f1:.3f} {'✅' if f1 >= 0.75 else '❌'}  |  AUC-ROC = {auc:.3f} {'✅' if auc > 0.80 else '❌'}")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
ConfusionMatrixDisplay.from_predictions(y_te, pred, ax=axes[0])
RocCurveDisplay.from_predictions(y_te, proba, ax=axes[1])
PrecisionRecallDisplay.from_predictions(y_te, proba, ax=axes[2])
plt.tight_layout()
```
