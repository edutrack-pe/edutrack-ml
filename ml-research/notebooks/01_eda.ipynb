```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
FEATURES = [
    "promedio", "inasistencias", "conducta", "horas_estudio", "nivel_edu_padres",
    "acceso_tecnologia", "tiempo_disponible_padres", "comunicacion_familiar",
    "cursos_desaprobados", "tareas_entregadas_pct",
]
TARGET = "en_riesgo"
```

```python
df = pd.read_csv("../data/dataset.csv")
df.head()
```

```python
df.info()
df.describe().T
```

## Balance de clases
Se espera que la clase `en_riesgo = 1` sea minoritaria (justifica SMOTE).

```python
df[TARGET].value_counts(normalize=True).plot.bar(title="Distribución de la clase")
```

```python
fig, axes = plt.subplots(2, 5, figsize=(20, 7))
for ax, col in zip(axes.flat, FEATURES):
    sns.boxplot(data=df, x=TARGET, y=col, ax=ax)
plt.tight_layout()
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df[FEATURES + [TARGET]].corr(), annot=True, fmt=".2f", cmap="RdBu_r", center=0)
```
